```python
import seaborn as sns
import pandas as pd

gr = sns.load_dataset('geyser')
gr.head()
```

```python
gr.groupby('kind').mean()
```

```python
import matplotlib.pyplot as plt

sns.pairplot(gr, hue='kind')
```

```python
sns.histplot(gr.waiting, kde=True)
```

```python
sns.jointplot(data=gr, x="waiting", y="duration", hue="kind", kind="kde")
```

```python
sns.scatterplot(data=gr, x="duration", y="waiting", hue="kind")
```
